```python
reviews_list = ['The food was really good and tasty.', 'Terrible service, I hated it.', 'A very good experience overall.', 'Not bad, but could be better.', 'Good ambiance and polite staff.', 'The worst food I have ever tasted.', 'It was a good meal for the price.', 'I will never order from here again.', 'Very good portion sizes.', 'Absolutely disgusting and cold.']
```

```python
def batch_predict_reviews(reviews):
    results = []
    for review in reviews:
        if 'good' in review.lower():
            results.append('positive')
        else:
            results.append('negative')
    return results
```

```python
predictions = batch_predict_reviews(reviews_list)
for i in range(len(reviews_list)):
    print(f'Review: {predictions[i]}')
```
